```python
###Comparison Between Trapezoidal,Simpson's 1/3rd and Gauss Legendre Methods

##Trapezoidal
def func(x):
    return (1/(x**2 + 1))
def trp_int(n,a,b):
    h = (b-a)/n
    P1 = func(a) + func(b)
    P2 = 0
    for i in range(1,n,2):
        s = (func(a + i*h))
        P2 = P2 + s
    P = (h/2)*(P1 + P2)
    return P
n = 40
a,b = 0,100
Result = trp_int(n,a,b)
print("Trapezoidal:", Result)







##Simpson's 1/3rd
def func(x):
    return (1/(x**2 + 1))
def smp_int(n,a,b):
    if n%2!= 0:
        print("Error: n must be even")
        return None
    h = (b-a)/n
    A = func(a) + func(b)
    Y = 0
    for i in range(1,n,2):
        B = 4*(func(a + i*h))
        Y = Y + B

    Z = 0
    for i in range(2,n,2):
        C = 2*(func(a + i*h))
        Z = Z + C
    R = (h/3)*(A+Y+Z)
    return R
n = 100
a,b = 0,100
Result = smp_int(n,a,b)
print("Simpsons 1/3rd:", Result)

##Gauss-Legendre Method
import numpy as np
def func(x):
  return (1/(x**2 + 1))
w = np.array([0.347855,0.652145,0.652145,0.347855])
x = np.array([-0.861136,-0.339981,0.339881,0.861136])

I = np.sum(w*func(x))

print("Gauss Legendre:", I)


```
