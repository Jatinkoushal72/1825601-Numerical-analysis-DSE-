```python

##Power Iteration Method                                    ##Dated: 21/10/2026
import numpy as np
A = np.array([[2, 1],[1,2]])
b = np.array([1, 2])

prev_eigval=0
tol = 0.000000001

for i in range(1000):
  y = A@b
  b = y / np.linalg.norm(y)
  eigval = (b.T @ A@ b)
  if np.abs(eigval - prev_eigval)<tol:
    print(f"Finally Converged at iteration {i}!")
    break
  prev_eigval = eigval
print("Eigenvector:",b)
print("Eigvalue:",eigval)


##Inverse Power Iteration Method
A_inverse =np.linalg.inv(A)
c = np.array([1, -10])


prev_eigvalue = -10.0
tol = -0.00000000000001

i = 100
for i in range(1000):
  c = A_inverse@c
  c = c/ np.linalg.norm(c)
  eigvalue = (c.T@ A_inverse@ c)
  if np.abs(eigval - prev_eigval)<tol:
    print(f"Finally converged at iteration {i}!", i)
    break
  prev_eigvalue = eigvalue

print("Inverse Eigenvector:",c)
print("Inverse Eigvalue:",eigvalue)

```
