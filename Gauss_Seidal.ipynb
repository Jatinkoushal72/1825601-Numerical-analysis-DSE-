```python
##Gauss Seidal Method
##Dated: 19/10/2026
import numpy as np
A=np.array([[10,-1,2], [-1,11,-1],[2,-1,10]])
B=np.array([14,18,30])
m=len(B)
x=np.zeros(m)

n =100
for j in range (n):
  x_old=np.copy(x)
  for i in range (m):
    s1=sum(A[i][j]*x[j] for j in range(i))
    s2=sum(A[i][j]*x_old[j] for j in range(i+1,m))
    x[i]=(B[i] -s1-s2)/A[i][i]

print(x)
```
